### Living Room

```python
jpeg = Image.open(
    os.path.join(JPEGImages, "living_room.jpg")
).convert("RGBA")
display(jpeg)
```

```python
segmentation = Image.open(
    os.path.join(SegmentationObject, "living_room.png")
).convert("RGBA")

display(segmentation)

alpha = segmentation.getchannel("A")
alpha = alpha.point(lambda p: p // 2)
segmentation.putalpha(alpha)
```

```python
overlay = Image.alpha_composite(jpeg, segmentation)
display(overlay)
```
